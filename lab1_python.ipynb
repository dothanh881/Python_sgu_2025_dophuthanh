```python
def isPrime(n):
  if n < 2:
    return False
  if n == 2:
    return True
  for i in range(2,n):
    if n % i == 0:
      return False
  return True
```

```python
def disNumberandSum(n):
  sum = 0
  for i in range(1,n):
    sum += i
    print(i)
  return f"sum: {sum}"
```

```python
print(disNumberandSum(10))
```

```python
def displayPrime(n):
  for i in range(1,n):
    if isPrime(i):
      print(i)
```

```python
displayPrime(10)
```

```python
def displayPrimeNFirst(n):
    count = 0
    num = 2
    while count < n:
        if isPrime(num):
            print(num)
            count += 1
        num += 1
```

```python
displayPrimeNFirst(10)
```

```python
def displayArray(arr, n):
    sum = 0
    for i in range(n):
        print(arr[i])
        sum += arr[i]
    print(f"sum: {sum}")
```

```python
arr = [10, 20, 30, 40, 50]
displayArray(arr, len(arr))
```

```python
def displayArrayPrime(arr, n):
    for i in range(n):
        if isPrime(arr[i]):
          print(arr[i])
```

```python
arr = [7, 2, 3, 4, 5]
displayArrayPrime(arr, len(arr))
```

```python
def addArray(arr,n):
  return arr.append(n)
def addArrayAtIndex(arr, index, value):
  arr.insert(index, value)
  return arr
def removeArray(arr, index):
  return arr.pop(index)
```

```python
arr = [10, 20, 30, 40, 50]
addArray(arr,3)
addArrayAtIndex(arr,3,4)
removeArray(arr, 0)
displayArray(arr, len(arr))
```

```python
def checkIsExist(arr, x):
  for i in range(len(arr)):
    if arr[i] == x:
      return i
  return False
```

```python
arr = [10, 20, 30, 40, 50]
checkIsExist(arr,3)
```
