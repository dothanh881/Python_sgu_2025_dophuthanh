***I. Vector***


Numpy (Numerical Python) - là một thư viện bao gồm các đối tượng mảng đa chiều (multidimentional array) và một tập hợp các phương thức để có thể xử lý mảng đó. Với numpy ta có thể sử dụng sụng các toán tử toán học, logic cùng các phương thức xử lý mảng để thực hiện trên đối tượng mục tiêu.
![numpy](https://i.imgur.com/mg8O3kd.png)

Một số ưu điểm của numpy:
- Chứa bộ các giá trị.
- Có thể chứa các kiểu dữ liệu khác nhau.
- Có thể thêm, xóa, sửa.
- Có đầy đủ chức năng tính toán.



Vector là một tuple của một hoặc nhiều giá trị vô hướng. Các vector được xây dựng từ các thành phần là các ordinary number (các kiểu int, float, ...). Có thể xem vector như một danh sách các số và đại số vector là các phép toán thực hiện trên các số bên trong danh sách đó.

Tên một vector là chữ thường - được ký hiệu như sau:
$$
  v = (x_1, x_2, x_3, ..., x_n)
$$
Trong đó:
- v là vector.
- $x_1, x_2, x_3, ..., x_n$ là các giá trị vô hướng

Một vector có thể được biểu diễn theo chiều ngang hoặc chiều dọc
$$
\begin{bmatrix}
   x_1 \\
   x_2 \\
   x_3 \\
   ... \\
   x_n
\end{bmatrix}
$$

```python
import numpy as np

# Tạo vector ngẫu nhiên
v1 = np.random.randint(1, 10, 5)  # vector 5 phần tử
v2 = np.random.randint(1, 10, 5)

print("Vector v1:", v1)
print("Vector v2:", v2)

```

### Cộng hai vector
$$
c = a + b
$$
$$
c = (a_1+b_1, a_2+b_2, ..., a_n+b_n)
$$

```python
# 1. Cộng vector
print("v1 + v2 =", v1 + v2)

```

### Trừ hai vector
$$
c = a - b
$$
$$
c = (a_1-b_1, a_2-b_2, ..., a_n-b_n)
$$

```python
# 2. Trừ vector
print("v1 - v2 =", v1 - v2)

```

# Nhân hai vector
$$
c = a * b
$$
$$
c = (a_1*b_1, a_2*b_2, ..., a_n*b_n)
$$

```python


# 3. Nhân từng phần tử (Hadamard product)
print("v1 * v2 =", v1 * v2)
```

# Chia hai vector
$$
c = a / b
$$
$$
c = (a_1/b_1, a_2/b_2, ..., a_n/b_n)
$$

```python
# 4. Chia từng phần tử
print("v1 / v2 =", v1 / v2)
```

```python

# 5. Nhân vô hướng
print("v1 . v2 =", np.dot(v1, v2))



# 6. Bình phương từng phần tử
print("v1^2 =", v1**2)
```

```python
# 7. Nối 2 vector
print("Concat [v1, v2] =", np.concatenate((v1, v2)))


```

```python
# 8. Cắt 1 đoạn vector (slice)
print("v1[1:4] =", v1[1:4])
```

```python
# 9. Lấy giá trị max/min
print("Max v1 =", np.max(v1), ", Min v1 =", np.min(v1))
```

```python
# 10. Sắp xếp vector
print("Sorted v1 =", np.sort(v1))
```

***II. Matrix***

```python
# Tạo ma trận
A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))

print("Ma trận A:\n", A)
print("Ma trận B:\n", B)
```

```python
# 1. Cộng ma trận
print("A + B =\n", A + B)

# 2. Trừ ma trận
print("A - B =\n", A - B)

# 3. Nhân từng phần tử
print("A * B =\n", A * B)

# 4. Tích ma trận (matrix multiplication)
print("A @ B =\n", A @ B)

# 5. Chuyển vị ma trận
print("A^T =\n", A.T)

```

```python
# 6. Nối ma trận theo hàng
print("Nối theo hàng =\n", np.hstack((A, B)))

# 7. Nối ma trận theo cột
print("Nối theo cột =\n", np.vstack((A, B)))

```

```python
# 8. Trích một dòng
print("Dòng 1 của A =", A[0])

# 9. Trích một cột
print("Cột 2 của A =", A[:, 1])

# 10. Cắt một khối con (sub-matrix)
print("Sub-matrix (2x2 từ A) =\n", A[:2, :2])

```

```python
# 11. Định thức ma trận
print("det(A) =", np.linalg.det(A))

# 12. Nghịch đảo ma trận (nếu khả nghịch)
try:
    print("A^-1 =\n", np.linalg.inv(A))
except np.linalg.LinAlgError:
    print("A không khả nghịch")
```
