### 1) Tải dữ liệu và tạo DataFrame
Dữ liệu ở dạng `Bunch` (tương tự dict). Ta tạo DataFrame từ `data` và `feature_names`, sau đó thêm cột `target` (0 = ác tính, 1 = lành tính).

```python
import pandas as pd
import sys, subprocess
from sklearn.datasets import load_breast_cancer

# Tải dataset
cancer_dataset = load_breast_cancer()

# Tạo DataFrame từ dữ liệu và tên cột
df = pd.DataFrame(cancer_dataset.data, columns=cancer_dataset.feature_names)

# Thêm cột 'target' (mục tiêu) vào DataFrame
# 0 = ác tính (malignant), 1 = lành tính (benign)
df['target'] = cancer_dataset.target
```

### 2) Xem nhanh dữ liệu: `head`, `tail`, `sample`
- `head(n)`: xem `n` dòng đầu để kiểm tra cấu trúc.
- `tail(n)`: xem `n` dòng cuối để đối chiếu dữ liệu toàn bộ.
- `sample(n)`: lấy ngẫu nhiên `n` dòng để kiểm tra phân phối tổng quát.

```python
print("5 dòng đầu tiên:")
display(df.head(5))
print("\n3 dòng cuối cùng:")
display(df.tail(3))
print("\n5 dòng ngẫu nhiên:")
display(df.sample(5, random_state=42))
```

### 3) Thông tin tổng quan: `info`
Hiển thị số lượng dòng/điểm dữ liệu, số cột, kiểu dữ liệu từng cột và số giá trị không-null. Đây là bước kiểm tra chất lượng dữ liệu cơ bản.

```python
df.info()
```

### 4) Thống kê mô tả: `describe`
Cung cấp số đếm, trung bình, độ lệch chuẩn, min/max và các phân vị (25%, 50%, 75%) cho các cột số, giúp nắm nhanh phạm vi và độ biến thiên.

```python
display(df.describe())
```

### 5) Phân bố nhãn: `value_counts`
Đếm số lượng từng nhãn trong cột phân loại, ở đây là `target` (0/1). Dùng để kiểm tra dữ liệu có cân bằng hay không.

```python
print("Phân bổ của target:")
print(df['target'].value_counts())
```

### 6) Số lượng giá trị duy nhất: `nunique`
`nunique()` giúp đếm số lượng giá trị khác nhau trong mỗi cột hoặc một cột cụ thể, hữu ích để đánh giá đa dạng dữ liệu.

```python
print("Số giá trị duy nhất theo cột:")
display(df.nunique())
print("\nSố giá trị duy nhất của 'mean radius':")
print(df['mean radius'].nunique())
```

### 7) Kiểm tra thiếu dữ liệu: `isnull` + `sum`
Đếm số giá trị thiếu (NaN) trên mỗi cột để ra quyết định xử lý. Bộ dữ liệu này thường không có thiếu, nhưng ta vẫn kiểm tra.

```python
missing = df.isnull().sum()
print("Số lượng giá trị thiếu theo cột:")
display(missing[missing > 0])
if missing.sum() == 0:
    print("Không có giá trị thiếu trong bộ dữ liệu này.")
```

### 8) Xử lý thiếu dữ liệu: `dropna` và `fillna` (minh hoạ)
Vì dataset không thiếu, ta tạo bản sao có thiếu nhằm minh hoạ: 
- `dropna(subset=[...])`: bỏ các dòng có NaN ở cột chỉ định.
- `fillna(value)`: điền giá trị thay thế (ví dụ median).

```python
import numpy as np
df_missing = df.copy()
df_missing.loc[:4, 'mean radius'] = np.nan  # tạo NaN ở 5 dòng đầu
print("Số NaN sau khi tạo:")
print(df_missing['mean radius'].isnull().sum())

# dropna: bỏ dòng có NaN ở mean radius
dropped = df_missing.dropna(subset=['mean radius'])
print("Kích thước sau dropna:", dropped.shape)

# fillna: điền median cho NaN
median_radius = df['mean radius'].median()
filled = df_missing.copy()
filled['mean radius'] = filled['mean radius'].fillna(median_radius)
print("Số NaN sau fillna:", filled['mean radius'].isnull().sum())
display(filled.head())
```

### 9) Nhóm và tổng hợp: `groupby` + `mean`
So sánh trung bình các đặc trưng theo `target` (0/1) để thấy khác biệt giữa u ác và u lành.

```python
mean_by_target = df.groupby('target').mean()
display(mean_by_target[['mean radius', 'mean texture', 'mean perimeter']])
```

### 10) Sắp xếp dữ liệu: `sort_values`
Xếp theo một cột để quan sát các giá trị lớn/nhỏ và so sánh liên quan giữa các đặc trưng.

```python
df_sorted = df.sort_values(by='mean perimeter', ascending=False)
display(df_sorted[['mean perimeter', 'mean radius', 'target']].head(10))
```

### 11) Quan hệ giữa các đặc trưng: `corr`
Ma trận tương quan giúp xem mức độ quan hệ tuyến tính giữa các đặc trưng số.

```python
display(df[['mean radius', 'mean texture', 'mean perimeter']].corr())
```

### 12) Biến đổi có điều kiện: `apply`
Tạo cột phân loại dựa trên ngưỡng (ví dụ median của `mean radius`) để phân nhóm kích thước.

```python
radius_median = df['mean radius'].median()
df['radius_category'] = df['mean radius'].apply(lambda x: 'Large' if x > radius_median else 'Small')
print("Phân bố radius_category:")
print(df['radius_category'].value_counts())
display(df[['mean radius', 'radius_category', 'target']].head())
```

### 13) Thêm cột tính toán: `assign`
Thêm cột mới từ công thức (ví dụ tích `mean radius` × `mean perimeter`) để tạo đặc trưng dẫn xuất.

```python
df2 = df.assign(radius_perimeter_product = df['mean radius'] * df['mean perimeter'])
display(df2[['mean radius', 'mean perimeter', 'radius_perimeter_product']].head())
```

### 14) Đổi tên cột: `rename`
Chuẩn hoá tên cột để tiện dùng (ví dụ thay khoảng trắng bằng dấu gạch dưới).

```python
df_renamed = df.rename(columns={'mean radius': 'mean_radius'})
display(df_renamed[['mean_radius', 'mean texture', 'mean perimeter']].head())
```

### 15) Kiểm tra trùng lặp: `duplicated`
Xác định và đếm số dòng trùng lặp, hữu ích để làm sạch dữ liệu.

```python
print("Số dòng trùng lặp:", df.duplicated().sum())
```

### 16) Phân vị: `quantile`
Lấy các phân vị chính (25%, 50%, 75%) để hiểu phân phối một đặc trưng.

```python
display(df['mean radius'].quantile([0.25, 0.5, 0.75]))
```
